```python
# Task 2: Caesar Cipher

message = input("Message: ")
shift = int(input("Shift: "))

encrypted = ""

# Encryption
for char in message:
    if 'A' <= char <= 'Z':
        encrypted += chr((ord(char) - ord('A') + shift) % 26 + ord('A'))
    elif 'a' <= char <= 'z':
        encrypted += chr((ord(char) - ord('a') + shift) % 26 + ord('a'))
    else:
        encrypted += char

print("Encrypted:", encrypted)

decrypted = ""

# Decryption
for char in encrypted:
    if 'A' <= char <= 'Z':
        decrypted += chr((ord(char) - ord('A') - shift) % 26 + ord('A'))
    elif 'a' <= char <= 'z':
        decrypted += chr((ord(char) - ord('a') - shift) % 26 + ord('a'))
    else:
        decrypted += char

print("Decrypted:", decrypted)
```

```python

# Task 3: Number Analyzer

start = int(input("Start: "))
end = int(input("End: "))

print("Number | Prime | Perfect | Armstrong | Palindrome | Digit Sum | Digits | Binary")

for num in range(start, end + 1):
    # Prime number
    if num > 1:
        prime = True
        for i in range(2, num):
            if num % i == 0:
                prime = False
                break
    else:
        prime = False

    # Perfect number
    total = 0
    for i in range(1, num):
        if num > 0 and num % i == 0:
            total += i
    perfect = num > 0 and total == num

    # Digit calculations
    digits = str(num)
    digit_sum = sum(int(d) for d in digits if d.isdigit())
    digit_count = sum(1 for d in digits if d.isdigit())

    # Armstrong number
    digit_values = [int(d) for d in digits if d.isdigit()]
    power = len(digit_values)
    armstrong = (
        num >= 0
        and sum(d ** power for d in digit_values) == num
    )

    # Palindrome number
    palindrome = num >= 0 and digits == digits[::-1]

    # Display results
    print(
        f"{num} | "
        f"{'Yes' if prime else 'No'} | "
        f"{'Yes' if perfect else 'No'} | "
        f"{'Yes' if armstrong else 'No'} | "
        f"{'Yes' if palindrome else 'No'} | "
        f"{digit_sum} | {digit_count} | {bin(num)[2:] if num >= 0 else 'N/A'}"
    )
```

```python

# Task 4: Number System Converter + GCD/LCM

num1 = int(input("Number 1: "))
num2 = int(input("Number 2: "))

# Function to convert decimal to another number system
def convert(number, base):
    if number == 0:
        return "0"

    digits = "0123456789ABCDEF"
    result = ""

    while number > 0:
        remainder = number % base
        result = digits[remainder] + result
        number = number // base

    return result

# Convert Number 1
print("Binary       :", convert(num1, 2))
print("Octal        :", convert(num1, 8))
print("Hexadecimal  :", convert(num1, 16))

# Calculate GCD using iterative logic
a = num1
b = num2

while b != 0:
    remainder = a % b
    a = b
    b = remainder

gcd = a

# Calculate LCM
if num1 == 0 or num2 == 0:
    lcm = 0
else:
    lcm = abs(num1 * num2) // gcd

print("GCD          :", gcd)
print("LCM          :", lcm)
```

```python

# Task 5: Two-Divisor FizzBuzz

n = int(input("N: "))

divisor1 = int(input("Divisor 1: "))
word1 = input("Word 1: ")

divisor2 = int(input("Divisor 2: "))
word2 = input("Word 2: ")

for i in range(1, n + 1):

    # Check both divisors first
    if i % divisor1 == 0 and i % divisor2 == 0:
        print(word1 + word2)

    elif i % divisor1 == 0:
        print(word1)

    elif i % divisor2 == 0:
        print(word2)

    else:
        print(i)
```

```python

# Task 6: Student Mark Conversion & Result Analyzer

students = int(input("Students: "))

print("\nStudent\tTotal\tAverage\tGrade")

for i in range(students):
    name = input("\nStudent name: ")
    marks_input = input("Enter marks separated by spaces: ").split()

    try:
        marks = [float(mark) for mark in marks_input]

        if not marks or any(mark < 0 or mark > 100 for mark in marks):
            print(name, "Invalid marks")
            continue

        total = sum(marks)
        average = total / len(marks)

        if average >= 90:
            grade = "A+"
        elif average >= 80:
            grade = "A"
        elif average >= 70:
            grade = "B"
        elif average >= 60:
            grade = "C"
        elif average >= 50:
            grade = "D"
        else:
            grade = "F"

        print(f"{name}\t{total:.0f}\t{average:.2f}\t{grade}")

    except ValueError:
        print(name, "Invalid marks")
```

```python

# Task 7: Digital Payment Transaction Analyzer

n = int(input("Transactions: "))

successful = []

rejected = 0

for i in range(n):
    amount_input = input("Enter transaction amount: ")

    try:
        amount = float(amount_input)

        if amount < 0:
            print("Rejected: Negative amount")
            rejected += 1
        else:
            successful.append(amount)

    except ValueError:
        print("Rejected: Invalid amount")
        rejected += 1

print("\n--- Transaction Report ---")

print("Successful Transactions :", len(successful))

total = sum(successful)

print(f"Total Amount           : ₹{total:.2f}")

if successful:
    highest = max(successful)
    lowest = min(successful)
    average = total / len(successful)

    print(f"Highest Transaction    : ₹{highest:.2f}")
    print(f"Lowest Transaction     : ₹{lowest:.2f}")
    print(f"Average Transaction    : ₹{average:.2f}")
else:
    print("No successful transactions.")

print("Rejected Transactions  :", rejected)
```

```python

# Task 8: String to Integer (atoi)

s = input("Enter a string: ")

i = 0
n = len(s)

# Skip leading spaces
while i < n and s[i] == ' ':
    i += 1

# Check the sign
sign = 1

if i < n and s[i] == '-':
    sign = -1
    i += 1
elif i < n and s[i] == '+':
    i += 1

# Convert digits into an integer
number = 0
found_digit = False

while i < n and '0' <= s[i] <= '9':
    number = number * 10 + (ord(s[i]) - ord('0'))
    found_digit = True
    i += 1

# Display the result
if found_digit:
    print(sign * number)
else:
    print(0)
```

```python

# Task 9: Add Binary

a = input("Enter first binary number: ")
b = input("Enter second binary number: ")

i = len(a) - 1
j = len(b) - 1

carry = 0
result = ""

# Add binary digits from right to left
while i >= 0 or j >= 0 or carry:

    digit1 = int(a[i]) if i >= 0 else 0
    digit2 = int(b[j]) if j >= 0 else 0

    total = digit1 + digit2 + carry

    result = str(total % 2) + result
    carry = total // 2

    i -= 1
    j -= 1

print("Binary Sum:", result)
```

```python

# Task 10: Happy Number

num = int(input("Enter a positive integer: "))

seen = set()

while num != 1 and num not in seen:
    seen.add(num)

    total = 0

    while num > 0:
        digit = num % 10
        total += digit ** 2
        num //= 10

    num = total

if num == 1:
    print(True)
else:
    print(False)
```
