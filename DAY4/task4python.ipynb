```python
n = int( input())
a = list(map(int, input().split()))
positive = negative = zero = 0
even = odd = 0
both = only3 = only5 = neither = 0
for x in a:
    if x > 0:
        positive += 1
    elif x < 0:
        negative += 1
    else:
        zero += 1
    if x != 0:
        if x % 2 ==0:
            even += 1
        else:
            odd += 1
    if x % 3 == 0 and x % 5 == 0:
        both += 1
    elif x % 3 == 0:
        only3 += 1
    elif x % 5 == 0:
        only5 +=  1
    else:
        neither += 1
print("positive:",positive)
print("negative:",negative)
print("zero:",zero)
print("even:",even)
print("odd:",odd)
print("Divisible by both 3 and 5:",both)
print("Divisible only by 3:",only3)
print("Divisible only by 5:",only5)
print("Divisible by neither:",neither)
```

```python
while True:
    print("1. celsius to fahrenheit")
    print("2. kilometers to miles")
    print("3. kilograms to pounds")
    print("4. exit")

    choice = int(input())

    if choice == 1:
        c = float(input())
        f = (c * 9/5) + 32
        print("fahrenheit:",f)

    elif choice == 2:
        km = float(input())
        miles = km * 0.6214
        print("kilometers to miles:",round(miles, 2))

    elif choice == 3:
        kg = float(input())
        pounds = kg * 2.20462
        print("kilograms to pounds:",round(pounds, 2))

    elif choice == 4:
        print("exit")
        break

    else:
        print("Invalid choice")
```

```python
n = int(input())
skipped = 0
for i in range(1,21):
    result = n* i
    if result > 100:
        break
    if result % 3 == 0:
        skipped += 1
        continue
    print(n, "x", i, "=", result)
print("skipped results:",skipped)
```

```python
p = input()
upper = lower = digit = special = False
for ch in p:
    if ch.isupper():
        upper = True
    elif ch.islower():
        lower = True
    elif ch.isdigit():
        digit = True
    else:
        special = True
if len(p) >= 8 and upper and lower and digit and special:
    print("Vaalid password")
else:
    print("Invalid password")
print("uppercase:", "present" if upper else "missing")
print("lowercase:", "present" if lower else "missing")
print("digit:", "present" if digit else "missing")
print("special character:", "present" if special  else "missing")
print("minimum length:", "satisfied" if len(p) >= 8 else "not satisfied")
```

```python
n = int(input())
for i in range(n):
    for j in range(n):
        if i == 0 or i == n-1 or j == 0 or j == n-1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
            
```

```python
L = int(input())
R = int(input())
found = False
for i in range(L, R + 1):
    j = 1
    while j * j < i :
        j += 1
    if j * j == i:
        print("first perfect square:", i)
        found = True
        break
if not found:
    print("no perfect square found")
```

```python

secret_number = 7
attempts = 5

for i in range(attempts):
    guess = int(input("Enter your guess: "))

    if guess < secret_number:
        print("Too Low")
    elif guess > secret_number:
        print("Too High")
    else:
        print("Correct")
        break
else:
    print("Game Over")
```

```python

normal = 0
large = 0
suspicious = 0
total = 0

while True:
    amount = int(input("Enter transaction amount: "))

    if amount == -1:
        break

    if amount == 0:
        pass
    elif amount < 0:
        continue
    elif amount <= 1000:
        normal += 1
        total += amount
    elif amount <= 5000:
        large += 1
        total += amount
    else:
        suspicious += 1
        total += amount

        if suspicious == 3:
            break

print("Normal transactions:", normal)
print("Large transactions:", large)
print("Suspicious transactions:", suspicious)
print("Total valid amount:", total)
```

```python

n = int(input("Enter a positive integer: "))
seen = set()

while n != 1 and n not in seen:
    seen.add(n)
    total = 0

    while n > 0:
        digit = n % 10
        total += digit ** 2
        n //= 10

    n = total

if n == 1:
    print(True)
else:
    print(False)


```
