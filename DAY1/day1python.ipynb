```python

# Task 1: Temperature Conversion & Formatted Output

celsius = float(input("Enter temperature in Celsius: "))

fahrenheit = (celsius * 9/5) + 32
kelvin = celsius + 273.15

print(f"Celsius: {celsius:.2f}°C")
print(f"Fahrenheit: {fahrenheit:.2f}°F")
print(f"Kelvin: {kelvin:.2f} K")

print("\nConversion Table")
print(f"{'Celsius':<12}{'Fahrenheit':<15}{'Kelvin':<10}")

for c in range(-40, 101, 10):
    f = (c * 9/5) + 32
    k = c + 273.15
    print(f"{c:<12.2f}{f:<15.2f}{k:<10.2f}")
```

```python

# Task 2: Three-Number Analyzer

# Read three numbers
a, b, c = map(int, input("Enter three numbers: ").split())

# Find largest, smallest, and average
print("Largest:", max(a, b, c))
print("Smallest:", min(a, b, c))
print(f"Average: {(a + b + c) / 3:.2f}")

# Read a separate number
n = int(input("Enter another number: "))

# Classify the number
if n > 0:
    sign = "Positive"
elif n < 0:
    sign = "Negative"
else:
    sign = "Zero"

if n % 2 == 0:
    parity = "Even"
else:
    parity = "Odd"

print(f"Classification: {sign} and {parity}")
```

```python

# Task 3: Leap Year + Grade Calculator

# Check leap year
year = int(input("Enter year: "))

if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print(f"{year} is a Leap Year")
else:
    print(f"{year} is not a Leap Year")

# Calculate grade
marks = int(input("Enter marks: "))

if marks < 0 or marks > 100:
    print("Invalid marks. Enter a value between 0 and 100.")
elif marks >= 90:
    print("Grade: A+")
elif marks >= 80:
    print("Grade: A")
elif marks >= 70:
    print("Grade: B")
elif marks >= 60:
    print("Grade: C")
elif marks >= 50:
    print("Grade: D")
else:
    print("Grade: F")
```

```python

# Task 4: Placement Bill Calculator

items = []

while True:
    price = input("Enter item price (or 'done'): ")

    if price.lower() == "done":
        break

    price = float(price)

    if price < 0:
        print("Price cannot be negative.")
        continue

    items.append(price)

# Calculate bill
subtotal = sum(items)

if subtotal >= 500:
    discount = subtotal * 0.10
else:
    discount = 0

tax = (subtotal - discount) * 0.05
final_amount = subtotal - discount + tax

# Print receipt
print("\nItem         Price")
print("-------------------")

for i, price in enumerate(items, start=1):
    print(f"Item {i:<6} {price:>8.2f}")

print("-------------------")
print(f"{'Subtotal':<15}{subtotal:>8.2f}")
print(f"{'Discount':<15}{discount:>8.2f}")
print(f"{'Tax':<15}{tax:>8.2f}")
print(f"{'Final Amount':<15}{final_amount:>8.2f}")
```

```python

# Task 6: Simple Interest Report

principal = float(input("Enter principal: "))
rate = float(input("Enter rate of interest: "))
time = float(input("Enter time in years: "))

if principal < 0 or time < 0:
    print("Invalid input. Principal and time must be non-negative.")
else:
    simple_interest = (principal * rate * time) / 100
    total_amount = principal + simple_interest

    print(f"Principal       : {principal:.2f}")
    print(f"Rate            : {rate:.2f}%")
    print(f"Time            : {time:.2f} years")
    print(f"Simple Interest : {simple_interest:.2f}")
    print(f"Total Amount    : {total_amount:.2f}")
```

```python

# Task 7: Type Inspector

value = input("Enter a value: ")

print("Original value type :", type(value).__name__)

try:
    integer_value = int(value)
    print("Integer value       :", integer_value)
    print("Integer type        :", type(integer_value).__name__)
except ValueError:
    print("Invalid integer conversion")

try:
    float_value = float(value)
    print("Float value         :", float_value)
    print("Float type          :", type(float_value).__name__)
except ValueError:
    print("Invalid float conversion")
```
