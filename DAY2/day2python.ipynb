```python

# Task 1: Prime, Factorial, Fibonacci, Digit Sum & Reverse

N = int(input("Enter N: "))
number = int(input("Enter a number: "))

# 1. Print prime numbers up to N
print("Primes:", end=" ")
for num in range(2, N + 1):
    is_prime = True
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(num, end=" ")

# 2. Calculate factorial of N
factorial = 1
for i in range(1, N + 1):
    factorial *= i
print("\nFactorial:", factorial)

# 3. Print first N Fibonacci numbers
a, b = 0, 1
print("Fibonacci:", end=" ")
for i in range(N):
    print(a, end=" ")
    a, b = b, a + b

# 4. Calculate sum of digits
temp = abs(number)
digit_sum = 0
while temp > 0:
    digit_sum += temp % 10
    temp //= 10
print("\nDigit Sum:", digit_sum)

# 5. Reverse the number
temp = abs(number)
reverse = 0
while temp > 0:
    digit = temp % 10
    reverse = reverse * 10 + digit
    temp //= 10

if number < 0:
    reverse = -reverse

print("Reverse:", reverse)
```

```python

# Task 2: Patterns & Multiplication Grid

N = int(input("Enter N: "))

# 1. Right Triangle
print("\nRight Triangle:")
for i in range(1, N + 1):
    for j in range(i):
        print("*", end="")
    print()

# 2. Pyramid
print("\nPyramid:")
for i in range(1, N + 1):
    for j in range(N - i):
        print(" ", end="")
    for j in range(2 * i - 1):
        print("*", end="")
    print()

# 3. Number Triangle
print("\nNumber Triangle:")
for i in range(1, N + 1):
    for j in range(1, i + 1):
        print(j, end="")
    print()

# 4. Multiplication Grid (1 to 10)
print("\nMultiplication Grid:")
for i in range(1, 11):
    for j in range(1, 11):
        print(f"{i} x {j} = {i*j}", end="\t")
    print()
```

```python

# Task 3: String Analyzer

text = input("Enter a string: ")

# 1. Reverse using a loop
reverse = ""
for char in text:
    reverse = char + reverse

print("Loop Reverse:", reverse)

# 2. Reverse using slicing
print("Slice Reverse:", text[::-1])

# 3. Check palindrome (ignore case and spaces)
cleaned = text.replace(" ", "").lower()

if cleaned == cleaned[::-1]:
    print("Palindrome: Yes")
else:
    print("Palindrome: No")

# 4. Count vowels, consonants and digits
vowels = 0
consonants = 0
digits = 0

for char in text.lower():
    if char in "aeiou":
        vowels += 1
    elif char.isalpha():
        consonants += 1
    elif char.isdigit():
        digits += 1

print("Vowels:", vowels)
print("Consonants:", consonants)
print("Digits:", digits)
```

```python

# Task 4: Formatted Electricity Bill

customer = input("Enter customer name: ")
units = int(input("Enter units consumed: "))

# Calculate electricity bill using slab-based charges
if units <= 100:
    amount = units * 3
elif units <= 200:
    amount = (100 * 3) + (units - 100) * 5
elif units <= 300:
    amount = (100 * 3) + (100 * 5) + (units - 200) * 7
else:
    amount = (100 * 3) + (100 * 5) + (100 * 7) + (units - 300) * 10

# Print formatted bill
print("\n----- Electricity Bill -----")
print(f"Customer : {customer}")
print(f"Units    : {units}")
print(f"Amount   : ₹{amount:.2f}")
```

```python

# Task 5: ATM Cash Withdrawal & Denomination Breakdown

amount = int(input("Enter withdrawal amount: "))

if amount <= 0:
    print("Invalid amount")

elif amount > 20000:
    print("Withdrawal limit exceeded")

elif amount % 10 != 0:
    print("Amount must be a multiple of 10")

else:
    remaining = amount
    total_notes = 0

    # Calculate ₹500 notes
    notes500 = remaining // 500
    remaining = remaining % 500
    total_notes += notes500

    # Calculate ₹200 notes
    notes200 = remaining // 200
    remaining = remaining % 200
    total_notes += notes200

    # Calculate ₹100 notes
    notes100 = remaining // 100
    remaining = remaining % 100
    total_notes += notes100

    # Calculate ₹50 notes
    notes50 = remaining // 50
    remaining = remaining % 50
    total_notes += notes50

    # Calculate ₹10 notes
    notes10 = remaining // 10
    total_notes += notes10

    print("\n----- ATM Receipt -----")
    print(f"₹500 notes : {notes500}")
    print(f"₹200 notes : {notes200}")
    print(f"₹100 notes : {notes100}")
    print(f"₹50 notes  : {notes50}")
    print(f"₹10 notes  : {notes10}")
    print(f"Total Notes: {total_notes}")
    print(f"Amount     : ₹{amount}")
```

```python

# Task 6: Digital Parking Fee & Time Calculator

entry = input("Enter entry time (HH:MM): ")
exit_time = input("Enter exit time (HH:MM): ")

# Convert entry time into minutes
entry_hours = int(entry[:2])
entry_minutes = int(entry[3:])
entry_total = entry_hours * 60 + entry_minutes

# Convert exit time into minutes
exit_hours = int(exit_time[:2])
exit_minutes = int(exit_time[3:])
exit_total = exit_hours * 60 + exit_minutes

# Calculate duration, handling midnight
duration = exit_total - entry_total

if duration < 0:
    duration += 24 * 60

hours = duration // 60
minutes = duration % 60

# Calculate billable hours
billable_hours = (duration + 59) // 60

# Calculate parking fee
if duration == 0:
    fee = 0
elif billable_hours == 1:
    fee = 30
else:
    fee = 30 + (billable_hours - 1) * 20

# Display receipt
print("\n----- Parking Receipt -----")
print(f"Parking Duration: {hours} hours {minutes} minutes")
print(f"Billable Hours  : {billable_hours}")
print(f"Parking Fee     : ₹{fee}")
```

```python

# Task 7: Delivery Charge & Earnings Report

n = int(input("Enter number of deliveries: "))

total_distance = 0
total_earnings = 0

for i in range(1, n + 1):
    distance = float(input(f"Enter distance for delivery {i} (km): "))

    total_distance += distance

    # Calculate earnings for each delivery
    if distance <= 5:
        earnings = 40
    else:
        earnings = 40 + (distance - 5) * 8

    total_earnings += earnings

# Calculate average distance
if n > 0:
    average_distance = total_distance / n
else:
    average_distance = 0

# Print earnings report
print("\n----- Delivery Earnings Report -----")
print(f"Total Distance   : {total_distance:.2f} km")
print(f"Total Earnings   : ₹{total_earnings:.2f}")
print(f"Average Distance : {average_distance:.2f} km")
```

```python

# Task 8: LeetCode #7 — Reverse Integer

x = int(input("Enter an integer: "))

temp = abs(x)
reverse = 0

while temp > 0:
    digit = temp % 10
    reverse = reverse * 10 + digit
    temp = temp // 10

# Restore the negative sign if needed
if x < 0:
    reverse = -reverse

print("Reversed Integer:", reverse)
```

```python

# Task 9: LeetCode #9 — Palindrome Number

x = int(input("Enter an integer: "))

if x < 0:
    print(False)
else:
    original = x
    reverse = 0

    while x > 0:
        digit = x % 10
        reverse = reverse * 10 + digit
        x = x // 10

    if original == reverse:
        print(True)
    else:
        print(False)
```

```python

# Task 10: LeetCode #412 — Fizz Buzz

N = int(input("Enter N: "))

for i in range(1, N + 1):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```
